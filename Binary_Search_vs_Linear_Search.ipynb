# Linear Search

**Linear Search** is a simple searching algorithm that checks each element in an array one by one until the target element is found or the end of the array is reached.

## How It Works

1. **Start:** Begin at the first element of the array.
2. **Compare:** Compare the current element with the target value.
3. **Move:** If they are not equal, move to the next element.
4. **Stop:** Return the position when the target is found, or report that it is not present.

## Summary

* **Best Case:** O(1) — the target is found at the first position.
* **Average Case:** O(n) — the target is found somewhere in the middle.
* **Worst Case:** O(n) — the target is at the last position or is not present.
* **Space Complexity:** O(1) — no extra space is required.


```python
import random
import matplotlib.pyplot as plt


def linear_search(arr, key):
    comparisons = 0

    for i in range(len(arr)):
        comparisons += 1

        if arr[i] == key:
            return comparisons

    return comparisons


```

# Binary Search

**Binary Search** is an efficient searching algorithm that repeatedly divides a **sorted array** in half to find the target element.

## How It Works

1. **Start:** Set the left and right boundaries of the sorted array.
2. **Find Middle:** Calculate the middle element of the current search range.
3. **Compare:** Compare the middle element with the target value.
4. **Reduce:** If the target is smaller, search the left half. If it is larger, search the right half.
5. **Repeat:** Continue dividing the search range until the target is found or the range becomes empty.

## Summary

* **Requirement:** The array must be **sorted**.
* **Best Case:** O(1) — the target is found at the middle on the first comparison.
* **Average Case:** O(log n) — the search range is repeatedly divided in half.
* **Worst Case:** O(log n) — the target is found after repeatedly reducing the search range or is not present.
* **Space Complexity:** O(1) — for the iterative implementation.


```python
# =========================================================
# BINARY SEARCH
# =========================================================

def binary_search(arr, key):
    low = 0
    high = len(arr) - 1
    comparisons = 0

    while low <= high:
        comparisons += 1

        mid = (low + high) // 2

        if arr[mid] == key:
            return comparisons

        elif key < arr[mid]:
            high = mid - 1

        else:
            low = mid + 1

    return comparisons

```

# Best, Average and Worst Case for Linear Search

* **Best Case:** The target is the **first element** in the array.
* **Time Complexity:** O(1) (Constant Time)

* **Average Case:** The target is found **somewhere in the middle** of the array.
* **Time Complexity:** O(n) (Linear Time)

* **Worst Case:** The target is the **last element** or is **not present** in the array.
* **Time Complexity:** O(n) (Linear Time)

- **Sequential Search:** Linear Search checks elements one by one from the beginning.
- **Best Case:** Only one comparison is needed.
- **Average Case:** Approximately half of the elements are checked.
- **Worst Case:** Every element must be checked.


```python


def create_array(n):
    return random.sample(range(1, n * 10 + 1), n)

sizes = [1, 2, 4, 8, 16, 32, 64, 128]

linear_best = []
linear_average = []
linear_worst = []

for n in sizes:
    arr = create_array(n)

    key = arr[0]
    comparisons = linear_search(arr, key)
    linear_best.append(comparisons)

    total = 0
    trials = 100

    for _ in range(trials):
        key = random.choice(arr)
        total += linear_search(arr, key)

    linear_average.append(total / trials)

    key = -1
    comparisons = linear_search(arr, key)
    linear_worst.append(comparisons)

print("LINEAR SEARCH")
print("=" * 55)
print("Size\tBest\tAverage\tWorst")

for i in range(len(sizes)):
    print(
        sizes[i],
        "\t",
        round(linear_best[i], 2),
        "\t",
        round(linear_average[i], 2),
        "\t",
        round(linear_worst[i], 2)
    )
```

# Best, Average and Worst Case for Binary Search

* **Best Case:** The target is the **middle element** of the sorted array.
* **Time Complexity:** O(1) (Constant Time)

* **Average Case:** The target is found after **several divisions** of the search range.
* **Time Complexity:** O(log n) (Logarithmic Time)

* **Worst Case:** The target is found after repeatedly dividing the array, or is **not present**.
* **Time Complexity:** O(log n) (Logarithmic Time)

- **Divide and Conquer:** Binary Search repeatedly divides the sorted array into two halves.
- **Best Case:** The target is found in the first comparison.
- **Average Case:** Only a small portion of the array needs to be searched.
- **Worst Case:** The search continues until the remaining search range becomes empty or the target is found.


```python
binary_best = []
binary_average = []
binary_worst = []

for n in sizes:
    arr = create_array(n)
    sorted_arr = sorted(arr)

    key = sorted_arr[(n - 1) // 2]
    comparisons = binary_search(sorted_arr, key)
    binary_best.append(comparisons)

    total = 0
    trials = 100

    for _ in range(trials):
        key = random.choice(sorted_arr)
        total += binary_search(sorted_arr, key)

    binary_average.append(total / trials)

    key = -1
    comparisons = binary_search(sorted_arr, key)
    binary_worst.append(comparisons)

print("BINARY SEARCH")
print("=" * 55)
print("Size\tBest\tAverage\tWorst")

for i in range(len(sizes)):
    print(
        sizes[i],
        "\t",
        round(binary_best[i], 2),
        "\t",
        round(binary_average[i], 2),
        "\t",
        round(binary_worst[i], 2)
    )
```



## Worst Case Comparison

Both Linear Search and Binary Search have different worst-case performance:

- **Linear Search:** O(n)
  - May need to check **every element** in the array.
  - The worst case occurs when the target is the last element or is not present.
  - The number of comparisons grows directly with the size of the array.

- **Binary Search:** O(log n)
  - Repeatedly divides the sorted array into **two halves**.
  - The worst case occurs when the target is found after several divisions or is not present.
  - The number of comparisons grows logarithmically with the array size.

## Conclusion

Binary Search is significantly more efficient than Linear Search for **large sorted arrays**.

As the array size increases, Linear Search requires increasingly more comparisons, while Binary Search eliminates half of the remaining elements after each comparison.

Therefore, **Linear Search is simpler and works on unsorted arrays, while Binary Search is much faster but requires the array to be sorted.**
