# Insertion Sort

**Insertion Sort** is a simple and intuitive sorting algorithm that builds a final sorted array one item at a time, much like how you sort a hand of playing cards.

## How It Works
1. **Divide:** The array is virtually split into a **sorted** part (initially just the first element) and an **unsorted** part.
2. **Pick & Compare:** The algorithm takes the first element from the unsorted part and compares it backward against the elements in the sorted part.
3. **Insert:** It shifts all larger elements one position to the right and inserts the picked element into its correct sorted position.
4. **Repeat:** This process continues until the entire array is sorted.

## Quick Summary
* **Analogy:** Sorting playing cards in your hands—picking one card at a time and sliding it into its proper place among your already sorted cards.
* **Time Complexity:**
  * **Best Case:** $O(n)$ when the array is already sorted.
  * **Worst & Average Case:** $O(n^2)$ when the array is in reverse order or randomly arranged.
* **Space Complexity:** $O(1)$ because it sorts the array in place without requiring extra memory.

In this notebook, we will code  insertion sort in python with an extra variable that will keep track of the number of comparisons. Based on this, we will plot graphs to observe its Time Complexity.

```python
#Insertion Sort

def insertion_sort(lst):
    count=0
    n=len(lst)
    for i in range(1,n):
        count+=1
        temp=lst[i]
        j=i-1
        while j>=0 and temp<lst[j]:
            count+=1
            lst[j+1]=lst[j]
            j-=1
            count+=1
        if j>=0:
            count+=1
        lst[j+1]=temp
    return lst,count

#Insertion Sort tested on an array
lst=[3,7,1,5,9]
print(insertion_sort(lst))
```

# Best Case for Insertion Sort

* **Best Case Scenario:** When the input array is **already sorted** in the desired order.
* **Time Complexity:** O(n) (Linear Time)


- **Single Comparison per Element:** The algorithm iterates through the array from left to right. For each element, it compares it against the immediately preceding element in the sorted portion.
- **No Shifts Required:** Because the array is already sorted, the current element is always larger than or equal to the one before it.
- **Immediate Termination:** The inner loop responsible for shifting elements to the right finds the correct position immediately on the first check, resulting in **zero shifts** for every element.
- **Linear Pass:** Instead of nested loops doing quadratic work (O(n^2)), it only performs a single forward pass with a constant number of comparisons per step, reducing the total work to linear time.

```python
# Best case tested on arrays of various sizes

b1 = []
for i in range(0, 20):
    b1.append(1 + i)

b2 = []
for i in range(0, 40):
    b2.append(1 + i)

b3 = []
for i in range(0, 60):
    b3.append(1 + i)

b4 = []
for i in range(0, 80):
    b4.append(1 + i)

b5 = []
for i in range(0, 100):
    b5.append(1 + i)

arrays = [b1, b2, b3, b4, b5]
bcounts=[]
for arr in arrays:
    n = len(arr)
    elements, comparisons = insertion_sort(arr)
    bcounts.append(comparisons)
    print(f"Number of elements: {n}, Comparisons: {comparisons}")
```

# Worst Case for Insertion Sort

* **Worst Case Scenario:** When the input array is **sorted in reverse order** (e.g., descending order when you want ascending).
* **Time Complexity:** O(n^2) (Quadratic Time)
* **Maximum Comparisons:** For every new element picked from the unsorted portion, the algorithm must compare it backward against *every* single element currently in the sorted portion to find its correct spot.
* **Frequent Shifting:** Because the array is reverse-sorted, every single preceding element is larger than the current element, forcing the inner loop to shift *all* previously sorted elements one position to the right every single time.
* **Triangular Workload:** The first element requires 1 comparison/shift, the second requires 2, the third requires 3, and so on up to $n-1$ comparisons for the final element.
* **Quadratic Growth:** This creates an arithmetic progression sum of $1 + 2 + 3 + \dots + (n-1)$, which simplifies to $\frac{n(n-1)}{2}$, resulting in an $O(n^2)$ time complexity.


```python
#Worst case tested on arrays of various sizes

w1=[]
for i in range(20,0,-1):
  w1.append(i)

w2=[]
for i in range(40,0,-1):
  w2.append(i)

w3=[]
for i in range(60,0,-1):
  w3.append(i)

w4=[]
for i in range(80,0,-1):
  w4.append(i)

w5=[]
for i in range(100,0,-1):
  w5.append(i)

wcounts = []
arrays = [w1,w2,w3,w4,w5]

for arr in arrays:
    n = len(arr)
    elements, comparisons = insertion_sort(arr)
    wcounts.append(comparisons)
    print(f"Number of elements: {n}, Comparisons: {comparisons}")


```

# Average Case for Insertion Sort

* **Average Case Scenario:** When the input array is **randomly ordered** with elements in arbitrary positions.
* **Time Complexity:** O(n^2) (Quadratic Time)
* **Partial Comparisons:** For any given element in a randomly arranged array, it will, on average, need to be compared against and shifted past about half of the elements currently in the sorted portion.
* **Probabilistic Positioning:** Elements are equally likely to land anywhere in the sorted subarray, meaning the inner loop runs for roughly half its maximum possible length on average.
* **Proportional Scaling:** The expected number of comparisons and shifts is still proportional to $\frac{n(n-1)}{4}$ (half of the worst-case scenario).
* **Quadratic Growth:** Even though it does half the work of the absolute worst case, the asymptotic growth rate remains quadratic, resulting in an $O(n^2)$ time complexity.

```python
#Average case tested on arrays of various sizes

import random
a1=[]
for i in range(0,20):
  a1.append(random.randint(1,100))

a2=[]
for i in range(0,40):
  a2.append(random.randint(1,100))

a3=[]
for i in range(0,60):
  a3.append(random.randint(1,100))

a4=[]
for i in range(0,80):
  a4.append(random.randint(1,100))

a5=[]
for i in range(0,100):
  a5.append(random.randint(1,100))

acounts=[]
arrays = [a1,a2,a3,a4,a5]

for arr in arrays:
    n = len(arr)
    elements, comparisons = insertion_sort(arr)
    acounts.append(comparisons)
    print(f"Number of elements: {n}, Comparisons: {comparisons}")
```

# Plotting Graphs

With our code for insertion sort done and the data for best, average and worst case for various array sizes collected, we can now plot them into graphs for observation.


```python
#Insertion Sort Best Case Graph

import matplotlib.pyplot as plt

ns=[20,40,60,80,100]
plt.plot(ns,bcounts,label="Best Case")


plt.xlabel("Input Size")
plt.ylabel("Number of Steps")
plt.title("Insertion Sort Analysis")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Insertion Sort Average Case Graph

ns=[20,40,60,80,100]

plt.plot(ns,acounts,label="Average Case")

plt.xlabel("Input Size")
plt.ylabel("Number of Steps")
plt.title("Insertion Sort Analysis")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Insertion Sort Worst Case Graph

ns=[20,40,60,80,100]

plt.plot(ns,wcounts,label="Worst Case")


plt.xlabel("Input Size")
plt.ylabel("Number of Steps")
plt.title("Insertion Sort Analysis")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Insertion Sort Best, Average, Worst Case Comparison Graph


ns=[20,40,60,80,100]

plt.plot(ns,wcounts,label="Worst Case")
plt.plot(ns,acounts,label="Average Case")
plt.plot(ns,bcounts,label="Best Case")

plt.xlabel("Input Size")
plt.ylabel("Number of Steps")
plt.title("Insertion Sort Analysis")
plt.legend()
plt.grid(True)
plt.show()
```

# Insertion Sort Complexity Analysis

* **Best Case Behavior:** When the input array is already sorted, the algorithm runs in linear time $O(n)$, represented by a straight-line growth on the graph.
* **Average Case Behavior:** When elements are randomly ordered, the algorithm operates at an expected time complexity of $O(n^2)$, forming a parabolic curve roughly halfway between best and worst cases.
* **Worst Case Behavior:** When the array is reverse-sorted, the algorithm hits its maximum quadratic bound of $O(n^2)$, producing the steepest upward curve due to maximum shifts and comparisons.
* **Growth Rate Comparison:** The combined analysis clearly highlights how quadratic algorithms scale dramatically worse than linear ones as input size ($n$) grows past 60 and approaches 100.

---

Thus, we can say that Insertion sort is exceptionally efficient for small or nearly sorted datasets due to its linear best-case performance, but its $O(n^2)$ average and worst-case complexities make it less ideal for large, randomly ordered inputs.
